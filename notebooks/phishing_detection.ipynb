```python
# Group the dataset features according to how they can be obtained

feature_groups = {
    "URL_or_Domain_Based": [
        "UsingIP",
        "LongURL",
        "ShortURL",
        "Symbol@",
        "Redirecting//",
        "PrefixSuffix-",
        "SubDomains",
        "HTTPS",
        "DomainRegLen",
        "NonStdPort",
        "HTTPSDomainURL",
        "AbnormalURL"
    ],

    "Webpage_Based": [
        "Favicon",
        "RequestURL",
        "AnchorURL",
        "LinksInScriptTags",
        "ServerFormHandler",
        "InfoEmail",
        "WebsiteForwarding",
        "StatusBarCust",
        "DisableRightClick",
        "UsingPopupWindow",
        "IframeRedirection"
    ],

    "External_or_Advanced": [
        "AgeofDomain",
        "DNSRecording",
        "WebsiteTraffic",
        "PageRank",
        "GoogleIndex",
        "LinksPointingToPage",
        "StatsReport"
    ]
}

# Display the feature groups
for group, features in feature_groups.items():
    print(f"\n{group}:")
    for feature in features:
        print(" -", feature)
```

```python
# Define the practical URL/domain-based features
selected_candidates = [
    "UsingIP",
    "LongURL",
    "ShortURL",
    "Symbol@",
    "Redirecting//",
    "PrefixSuffix-",
    "SubDomains",
    "HTTPS",
    "DomainRegLen",
    "NonStdPort",
    "HTTPSDomainURL",
    "AbnormalURL"
]

```

```python
# Check how strongly each candidate feature is correlated with the target class
candidate_correlations = (
    data[selected_candidates + ["class"]]
    .corr(numeric_only=True)["class"]
    .drop("class")
    .sort_values(key=abs, ascending=False)
)

candidate_correlations
```

```python
# ==========================================
# TRAIN-TEST SPLIT
# ==========================================

from sklearn.model_selection import train_test_split

# Separate the input features from the target
X = data.drop(columns=['class', 'Index'])
y = data['class']

# Split the dataset into 80% training and 20% testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the sizes of the resulting datasets
print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# ==========================================
# DECISION TREE
# ==========================================

from sklearn.tree import DecisionTreeClassifier

# Create the Decision Tree model
dt_model = DecisionTreeClassifier(
    random_state=42
)

# Train the model
dt_model.fit(X_train, y_train)
```

```python
# ==========================================
# RANDOM FOREST
# ==========================================

from sklearn.ensemble import RandomForestClassifier

# Create the Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model
rf_model.fit(X_train, y_train)
```

```python
# ==========================================
# SUPPORT VECTOR MACHINE
# ==========================================

from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

# Create a pipeline that scales the data before applying SVM
svm_model = make_pipeline(
    StandardScaler(),
    SVC()
)

# Train the model
svm_model.fit(X_train, y_train)
```

```python
# ==========================================
# MODEL PREDICTIONS
# ==========================================

# Generate predictions using the test data
dt_pred = dt_model.predict(X_test)
rf_pred = rf_model.predict(X_test)
svm_pred = svm_model.predict(X_test)
```

```python
# Display a few predictions from each model
print("Decision Tree:", dt_pred[:10])
print("Random Forest:", rf_pred[:10])
print("SVM:", svm_pred[:10])
```

```python
# ==========================================
# MODEL EVALUATION
# ==========================================

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Calculate performance metrics for each model
results = [
    {
        'Model': 'Decision Tree',
        'Accuracy': accuracy_score(y_test, dt_pred),
        'Precision': precision_score(y_test, dt_pred, pos_label=1),
        'Recall': recall_score(y_test, dt_pred, pos_label=1),
        'F1-Score': f1_score(y_test, dt_pred, pos_label=1)
    },
    {
        'Model': 'Random Forest',
        'Accuracy': accuracy_score(y_test, rf_pred),
        'Precision': precision_score(y_test, rf_pred, pos_label=1),
        'Recall': recall_score(y_test, rf_pred, pos_label=1),
        'F1-Score': f1_score(y_test, rf_pred, pos_label=1)
    },
    {
        'Model': 'SVM',
        'Accuracy': accuracy_score(y_test, svm_pred),
        'Precision': precision_score(y_test, svm_pred, pos_label=1),
        'Recall': recall_score(y_test, svm_pred, pos_label=1),
        'F1-Score': f1_score(y_test, svm_pred, pos_label=1)
    }
]

# Convert the results into a table
results_df = pd.DataFrame(results)

# Display the results
results_df
```

```python
# ==========================================
# RANDOM FOREST CONFUSION MATRIX
# ==========================================

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Calculate the confusion matrix
rf_cm = confusion_matrix(y_test, rf_pred)

# Display the confusion matrix
disp = ConfusionMatrixDisplay(
    confusion_matrix=rf_cm,
    display_labels=['-1', '1']
)

disp.plot()

plt.title('Random Forest Confusion Matrix')
plt.show()
```

```python
# ==========================================
# CONFUSION MATRICES FOR ALL MODELS
# ==========================================

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Decision Tree
ConfusionMatrixDisplay(
    confusion_matrix(y_test, dt_pred),
    display_labels=['-1', '1']
).plot(ax=axes[0], colorbar=False)
axes[0].set_title('Decision Tree')

# Random Forest
ConfusionMatrixDisplay(
    confusion_matrix(y_test, rf_pred),
    display_labels=['-1', '1']
).plot(ax=axes[1], colorbar=False)
axes[1].set_title('Random Forest')

# SVM
ConfusionMatrixDisplay(
    confusion_matrix(y_test, svm_pred),
    display_labels=['-1', '1']
).plot(ax=axes[2], colorbar=False)
axes[2].set_title('SVM')

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# RANDOM FOREST FEATURE IMPORTANCE
# ==========================================

# Get the importance of every feature
feature_importance = pd.Series(
    rf_model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

# Display the feature importance
feature_importance
```
